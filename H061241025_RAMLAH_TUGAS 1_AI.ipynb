## **A. Pendahuluan**

Dalam kegiatan geofisika seperti pengolahan data, pemodelan, dan interpretasi, prompt untuk AI tidak hanya berfungsi sebagai pertanyaan, tetapi juga sebagai petunjuk kerja yang terstruktur. Prompt yang disusun dengan baik membantu AI memberikan analisis yang lebih terarah serta memudahkan pengguna memahami dasar dari setiap proses yang dilakukan.

Struktur prompt dapat disusun menggunakan beberapa komponen utama yang membentuk alur kerja, yaitu **R-T-C-I-K-O-Q-V** (*Role, Task, Context, Input, Constraints, Output, Quality Criteria, dan Optional Verification*).

---

## **B. Penjelasan Tiap Komponen**

Framework **R-T-C-I-K-O-Q-V** terdiri atas beberapa bagian yang saling melengkapi untuk membentuk prompt yang terarah. Setiap komponen memiliki fungsi yang berbeda dalam membantu AI memahami tugas dan menghasilkan respons sesuai kebutuhan.

**2.1 Role (R)**

Menentukan **peran atau keahlian AI** dalam menjalankan tugas. Role membantu AI menyesuaikan sudut pandang dan pengetahuan yang digunakan sesuai dengan permasalahan yang diberikan.

**2.2 Task (T)**

Menjelaskan **tugas utama yang harus dilakukan oleh AI**. Bagian ini harus dibuat secara jelas agar AI memahami tujuan analisis yang ingin dicapai.

**2.3 Context (C)**

Memberikan **latar belakang atau kondisi permasalahan** yang sedang dianalisis. Context membantu AI memahami situasi penelitian sehingga analisis tidak dilakukan tanpa mempertimbangkan kondisi yang ada.

**2.4 Input (I)**

Berisi **data atau informasi yang diberikan kepada AI** sebagai dasar untuk melakukan analisis. Input dapat berupa data pengukuran, parameter, tabel, grafik, maupun informasi pendukung lainnya.

**2.5 Constraints (K)**

Menentukan **batasan dalam proses analisis**. Bagian ini digunakan untuk mencegah AI membuat asumsi atau memberikan informasi yang tidak didukung oleh data yang tersedia.

**2.6 Output (O)**

Menjelaskan **bentuk keluaran yang diharapkan dari AI**. Output dapat berupa narasi, tabel, interpretasi, langkah analisis, atau bentuk lainnya sesuai tujuan pekerjaan.

**2.7 Quality Criteria (Q)**

Berisi **kriteria yang digunakan untuk memastikan kualitas jawaban AI**. Kriteria ini dapat mencakup ketepatan konsep, kesesuaian dengan data, kejelasan penjelasan, dan konsistensi hasil analisis.

**2.8 Optional Verification (V)**

Merupakan bagian **pemeriksaan tambahan** yang dilakukan apabila diperlukan. Verifikasi dapat digunakan untuk membandingkan hasil analisis dengan data pendukung agar diperoleh hasil yang lebih konsisten.

## **C. Contoh Implementasi Prompt**

### **1. Konfigurasi Wenner–Schlumberger (Pemetaan Penampang 2D Resistivitas Regional)**

**Role**

Anda bertindak sebagai ahli geofisika yang memiliki spesialisasi dalam metode geolistrik resistivitas, geofisika lingkungan, dan hidrogeologi. Anda memahami prinsip konfigurasi Wenner–Schlumberger, perhitungan resistivitas semu, serta interpretasi penampang resistivitas 2D untuk mengidentifikasi indikasi pencemaran air lindi di Tempat Pembuangan Akhir (TPA) Tamangapa, Antang, Makassar.

**Output**

Menghasilkan analisis geofisika lingkungan yang mencakup perhitungan parameter kelistrikan, tabel hasil pengolahan data, visualisasi penampang resistivitas 2D, dan interpretasi zona konduktif berdasarkan variasi resistivitas bawah permukaan.

**Task**

Lakukan pengolahan dan interpretasi data geolistrik menggunakan konfigurasi Wenner–Schlumberger untuk memetakan distribusi resistivitas bawah permukaan di sekitar TPA Tamangapa.

Tahapan pengerjaan:

1. Hitung resistansi listrik menggunakan persamaan:
   $$R=\frac{V}{I}$$
2. Tentukan faktor geometri sesuai konfigurasi dan susunan elektroda yang digunakan.
3. Hitung resistivitas semu menggunakan persamaan:

   $$\rho_a=KR$$

4. Sajikan hasil perhitungan dalam tabel yang memuat spasi elektroda, faktor pengukuran, arus, beda potensial, resistansi, faktor geometri, dan resistivitas semu.
5. Buat visualisasi penampang resistivitas 2D yang menunjukkan variasi resistivitas terhadap jarak lintasan dan kedalaman.
6. Interpretasikan zona konduktif, zona transisi, dan zona yang lebih resistif.

**Context**

TPA Tamangapa berada di Kecamatan Manggala, kawasan Antang, Kota Makassar. Akumulasi sampah dan infiltrasi cairan hasil dekomposisi berpotensi menghasilkan air lindi yang memengaruhi sifat kelistrikan tanah.

Zona dengan resistivitas rendah dapat mengindikasikan material yang kaya fluida atau memiliki kandungan ion terlarut tinggi. Kondisi geologi regional berupa endapan aluvial yang menutupi batuan gunungapi menjadi informasi pendukung interpretasi.

**Input**

Gunakan data pengukuran dan parameter yang telah disediakan dalam notebook, termasuk spasi elektroda, faktor pengukuran, arus listrik, beda potensial, serta informasi geologi regional.

Jika data lapangan tidak tersedia, gunakan data simulasi yang telah disediakan dan nyatakan secara eksplisit bahwa hasilnya merupakan simulasi, bukan hasil survei lapangan terverifikasi.

**Constraints**

1. Terapkan rumus faktor geometri yang sesuai dengan konfigurasi dan definisi spasi elektroda yang digunakan.
2. Konversikan arus dari mA ke A dan beda potensial dari mV ke V sebelum menghitung resistansi.
3. Gunakan satuan SI secara konsisten dan cantumkan satuan pada tabel serta grafik.
4. Gunakan resistivitas di bawah $5\ \Omega\cdot\text{m}$ sebagai ambang indikatif zona yang sangat konduktif dan berpotensi berkaitan dengan air lindi, bukan sebagai bukti tunggal pencemaran.
5. Jangan menetapkan litologi maupun batas pencemaran secara mutlak tanpa data pendukung.
6. Gunakan Python dan pustaka NumPy serta Matplotlib untuk perhitungan dan visualisasi.

**Quality Criteria**

- Seluruh perhitungan harus konsisten secara matematis.
- Tabel harus memuat seluruh parameter yang diperlukan.
- Visualisasi harus memiliki judul, label sumbu, satuan, dan colorbar yang jelas.
- Interpretasi harus menghubungkan pola resistivitas dengan kemungkinan kondisi hidrogeologi secara logis.
- Kesimpulan tidak boleh melebihi kemampuan metode geolistrik.

**Output Akhir**

1. Tabel perhitungan resistansi, faktor geometri, dan resistivitas semu.
2. Visualisasi penampang resistivitas 2D bawah permukaan.
3. Interpretasi zona konduktif, zona transisi, dan zona resistif.
4. Kesimpulan mengenai indikasi sebaran air lindi berdasarkan model yang dianalisis.

## **2. Konfigurasi Wenner (Analisis Sebaran Lateral Dangkal)**

**Role**

Anda bertindak sebagai ahli geofisika lingkungan yang menguasai konfigurasi Wenner untuk investigasi resistivitas dangkal, pemetaan variasi lateral, dan identifikasi anomali konduktif yang berpotensi berhubungan dengan sebaran air lindi di TPA Tamangapa, Antang, Makassar.

**Output**

Menghasilkan tabel pengolahan data resistivitas konfigurasi Wenner, visualisasi kontur resistivitas 2D, dan interpretasi distribusi lateral anomali konduktif pada lapisan tanah dangkal.

**Task**

Olah data geolistrik konfigurasi Wenner dengan menghitung resistansi listrik, faktor geometri, dan resistivitas semu setiap pengukuran.

Gunakan persamaan:

$$R=\frac{V}{I}$$

$$\rho_a=KR$$

Tahapan pengerjaan:

1. Hitung resistansi berdasarkan arus dan beda potensial.
2. Tentukan faktor geometri konfigurasi Wenner.
3. Hitung resistivitas semu setiap pengukuran.
4. Susun hasil perhitungan secara sistematis dalam bentuk tabel.
5. Buat visualisasi kontur 2D yang menggambarkan variasi resistivitas terhadap jarak lintasan dan kedalaman estimasi.
6. Identifikasi zona dengan resistivitas rendah, zona transisi, dan zona dengan resistivitas relatif tinggi.
7. Bandingkan pola tersebut untuk mengevaluasi kemungkinan perubahan resistivitas secara lateral di sekitar area pembuangan sampah.

**Context**

Air lindi memiliki kandungan ion terlarut yang dapat meningkatkan konduktivitas listrik material tanah. Oleh karena itu, anomali resistivitas rendah di sekitar TPA Tamangapa dapat menjadi petunjuk awal keberadaan material jenuh fluida atau zona yang berpotensi terpengaruh lindi.

Konfigurasi Wenner digunakan untuk mengevaluasi perubahan resistivitas dangkal, terutama variasi horizontal, dengan tetap mempertimbangkan sensitivitas vertikalnya.

**Input**

Gunakan data konfigurasi Wenner yang telah tersedia dalam notebook, termasuk:

- Jarak antar-elektroda $a$.
- Arus listrik $I$.
- Beda potensial $V$.
- Koordinat atau posisi pengukuran, jika tersedia.

Gunakan data simulasi yang sudah ada apabila belum tersedia data lapangan terverifikasi. Pertahankan konsistensi data input dan jangan mengganti nilainya hanya untuk menghasilkan pola visual tertentu.

**Constraints**

1. Untuk konfigurasi Wenner dengan jarak antar-elektroda berurutan yang sama sebesar $a$, gunakan faktor geometri:

   $$K=2\pi a$$

2. Konversikan seluruh besaran listrik ke satuan SI sebelum perhitungan.
3. Hitung resistivitas semu berdasarkan nilai pengukuran yang tersedia.
4. Jangan menyamakan resistivitas semu dengan resistivitas sebenarnya tanpa proses inversi yang sesuai.
5. Gunakan ambang $\rho_a<5\ \Omega\cdot\text{m}$ sebagai indikator anomali konduktif kuat yang perlu diperiksa lebih lanjut, bukan sebagai konfirmasi tunggal pencemaran.
6. Buat grafik menggunakan Matplotlib dengan skala, legenda, label sumbu, dan colorbar yang jelas.

**Quality Criteria**

- Tabel perhitungan harus lengkap dan dapat ditelusuri kembali ke data input.
- Visualisasi harus memperlihatkan kontras anomali secara jelas tanpa menciptakan hasil pengukuran yang tidak tersedia.
- Interpretasi harus menjelaskan hubungan antara distribusi resistivitas dan kemungkinan kondisi tanah, fluida, atau kontaminasi secara hati-hati.
- Setiap grafik harus memiliki judul, label sumbu, dan satuan yang sesuai.

**Output Akhir**

1. Tabel hasil perhitungan konfigurasi Wenner.
2. Visualisasi kontur resistivitas 2D.
3. Analisis distribusi lateral zona konduktif dan zona resistivitas lebih tinggi.
4. Interpretasi geofisika lingkungan mengenai kemungkinan sebaran dangkal air lindi.

### **3. Konfigurasi Schlumberger / VES (Analisis Pelapisan Vertikal 1D)**

**Role**

Anda bertindak sebagai ahli geofisika eksplorasi dangkal dan hidrogeologi yang menguasai metode Vertical Electrical Sounding (VES) menggunakan konfigurasi Schlumberger. Anda mampu mengolah kurva resistivitas semu, mengevaluasi variasi resistivitas terhadap kedalaman, dan menyusun model interpretasi pelapisan bawah permukaan di kawasan TPA Tamangapa, Antang, Makassar.

**Output**

Menghasilkan tabel pengolahan data VES, kurva sounding log-log, model interpretasi pelapisan 1D, dan uraian mengenai kemungkinan hubungan antara zona konduktif, akuifer dangkal, dan batuan dasar.

**Task**

Lakukan pengolahan data VES konfigurasi Schlumberger secara berurutan.

1. Hitung resistansi listrik dari rasio beda potensial terhadap arus:

   $$R=\frac{V}{I}$$

2. Hitung faktor geometri berdasarkan jarak elektroda arus dan potensial sesuai definisi yang digunakan.
3. Tentukan resistivitas semu setiap pengukuran:

   $$\rho_a=KR$$

4. Buat kurva hubungan antara setengah bentangan elektroda arus $AB/2$ dan resistivitas semu $\rho_a$ menggunakan skala log-log.
5. Susun model pelapisan 1D yang memperlihatkan resistivitas dan ketebalan setiap lapisan berdasarkan parameter model yang tersedia.
6. Interpretasikan kemungkinan zona topsoil, material jenuh fluida, akuifer dangkal, dan batuan dasar dengan mempertimbangkan informasi geologi regional.

**Context**

Metode VES digunakan untuk mempelajari perubahan resistivitas terhadap kedalaman melalui perubahan bentangan elektroda.

Di sekitar TPA Tamangapa, anomali konduktif dapat berkaitan dengan tanah yang mengandung fluida kaya ion terlarut, sedangkan lapisan yang lebih resistif dapat mencerminkan kondisi material yang berbeda.

Interpretasi harus mempertimbangkan bahwa satu nilai resistivitas dapat mewakili beberapa jenis material dan bahwa model 1D merupakan penyederhanaan kondisi bawah permukaan.

**Input**

Gunakan data VES yang tersedia pada notebook, meliputi:

- Setengah bentangan elektroda arus $AB/2$.
- Setengah bentangan elektroda potensial $MN/2$.
- Arus injeksi.
- Beda potensial.
- Parameter resistivitas dan ketebalan lapisan jika tersedia.

Jika model pelapisan dengan nilai resistivitas dan ketebalan telah disediakan, gunakan nilai tersebut sebagai model ilustratif dan bedakan dari hasil inversi data pengukuran.

Jangan mengklaim bahwa ketebalan lapisan atau kedalaman akuifer telah terverifikasi jika tidak didukung data inversi atau pemboran.

**Constraints**

1. Gunakan faktor geometri Schlumberger yang sesuai dengan definisi jarak setengah bentangan $AB/2$ dan $MN/2$:

   $$K=\pi\frac{(AB/2)^2-(MN/2)^2}{2(MN/2)}$$

2. Konversikan arus dan beda potensial ke satuan SI sebelum menghitung resistansi.
3. Sajikan kurva VES dengan sumbu logaritmik dan satuan yang benar.
4. Bedakan secara jelas resistivitas semu hasil perhitungan dengan resistivitas sebenarnya pada model lapisan.
5. Jangan menyatakan bahwa keberadaan akuifer atau batuan basalt telah terbukti hanya berdasarkan resistivitas.
6. Gunakan Python, NumPy, dan Matplotlib untuk menghasilkan tabel dan grafik.

**Quality Criteria**

- Perhitungan harus konsisten dengan geometri elektroda.
- Kurva log-log harus terbaca dengan jelas.
- Model pelapisan harus menunjukkan nilai resistivitas serta kedalaman atau ketebalan yang jelas.
- Interpretasi harus menjelaskan dasar geofisikanya, keterbatasan metode VES, dan kebutuhan data pendukung untuk memastikan litologi serta kondisi akuifer.

**Output Akhir**

1. Tabel hasil pengolahan data VES.
2. Kurva sounding resistivitas semu log-log.
3. Model interpretasi pelapisan bawah permukaan 1D.
4. Interpretasi kemungkinan kedalaman zona konduktif, akuifer dangkal, dan batuan dasar.

### **4. Konfigurasi Dipole–Dipole (Analisis Sebaran Lateral dan Visualisasi 3D)**

**Role**

Anda bertindak sebagai ahli geofisika yang memiliki keahlian dalam metode resistivitas konfigurasi Dipole–Dipole, pemetaan anomali lateral, dan visualisasi numerik untuk investigasi geofisika lingkungan.

Anda memahami penggunaan faktor geometri, analisis resistivitas semu, dan keterbatasan interpretasi model 3D dalam kajian potensi migrasi air lindi di TPA Tamangapa, Antang, Makassar.

**Output**

Menghasilkan tabel perhitungan konfigurasi Dipole–Dipole, grafik resistivitas berdasarkan tingkat pemisahan elektroda, visualisasi permukaan 3D, serta interpretasi pola distribusi anomali konduktif yang berpotensi berhubungan dengan sebaran air lindi.

**Task**

Olah data konfigurasi Dipole–Dipole menggunakan nilai spasi dipol, faktor pemisahan $n$, arus listrik, dan beda potensial yang tersedia.

Tahapan pengerjaan:

1. Hitung resistansi listrik:

   $$R=\frac{V}{I}$$

2. Hitung faktor geometri sesuai konfigurasi Dipole–Dipole.
3. Hitung resistivitas semu setiap pengukuran:

   $$\rho_a=KR$$

4. Sajikan hasil pengolahan dalam tabel.
5. Visualisasikan perubahan resistivitas terhadap tingkat pemisahan elektroda.
6. Buat visualisasi 3D untuk memperlihatkan distribusi spasial anomali berdasarkan data atau model simulasi yang tersedia.
7. Interpretasikan zona dengan resistivitas rendah dan tinggi serta diskusikan kemungkinan jalur migrasi fluida berdasarkan hubungan anomali dengan kondisi geologi dan topografi jika informasi tersebut tersedia.

**Context**

Konfigurasi Dipole–Dipole memiliki sensitivitas yang baik terhadap perubahan resistivitas lateral dan struktur yang kompleks.

Dalam kajian TPA Tamangapa, anomali konduktif dapat digunakan sebagai petunjuk awal untuk menelusuri kemungkinan sebaran fluida yang berpotensi mengandung air lindi.

Apabila hanya tersedia data satu lintasan atau sejumlah kecil tingkat pemisahan, nyatakan keterbatasan cakupan spasial.

Jika visualisasi menggunakan fungsi matematis simulasi, tandai hasilnya sebagai ilustrasi model, bukan hasil pemetaan 3D lapangan.

**Constraints**

1. Gunakan faktor geometri Dipole–Dipole yang konsisten dengan definisi spasi dipol:

   $$K=\pi a n(n+1)(n+2)$$

3. Hitung resistivitas semu menggunakan $\rho_a=KR$.
4. Jangan menganggap tingkat pemisahan $n$ sebagai kedalaman investigasi yang pasti.
5. Jangan mengklaim arah migrasi air lindi hanya berdasarkan bentuk permukaan 3D tanpa data posisi, topografi, atau informasi hidrogeologi yang mendukung.

**Quality Criteria**

- Seluruh parameter harus dihitung dengan rumus yang tepat.
- Visualisasi harus konsisten dengan data yang digunakan.
- Perbedaan antara data pengukuran dengan simulasi harus dijelaskan.
- Interpretasi harus membahas distribusi lateral anomali, kemungkinan kontrol geologi, serta keterbatasan dalam menentukan jalur migrasi lindi.

**Output Akhir**

1. Tabel pengolahan data Dipole–Dipole.
2. Grafik resistivitas terhadap faktor pemisahan elektroda.
3. Visualisasi permukaan 3D berdasarkan data atau simulasi yang tersedia.
4. Interpretasi pola spasial anomali konduktif dan keterbatasan penentuan arah migrasi lindi.

### **5. Validasi Silang (Cross-Validation) dan Integrasi Hasil Geolistrik**

**Role**

Anda bertindak sebagai ahli geofisika lingkungan yang menguasai evaluasi kualitas data, analisis statistik, validasi model resistivitas, serta integrasi hasil metode Wenner–Schlumberger, Wenner, Schlumberger/VES, dan Dipole–Dipole.

Anda mampu mengevaluasi kesesuaian antara data pengukuran dan model serta menyusun interpretasi terpadu untuk kajian potensi pencemaran air lindi di TPA Tamangapa, Antang, Makassar.

**Output**

Menghasilkan tabel evaluasi model, grafik perbandingan resistivitas pengukuran dan model, analisis residual, serta kesimpulan terpadu mengenai konsistensi anomali geolistrik dan keterbatasan interpretasi hidrogeologi.

**Task**

Integrasikan hasil pengolahan dari empat metode geolistrik yang telah dilakukan sebelumnya.

Tahapan pengerjaan:

1. Bandingkan nilai resistivitas pengukuran dengan nilai resistivitas model pada pasangan titik yang sesuai.
2. Hitung residual menggunakan persamaan:

   $$e_i=\rho_{i,\mathrm{ukur}}-\rho_{i,\mathrm{model}}$$

3. Hitung Root Mean Square Error (RMSE) apabila data yang tersedia memenuhi persyaratan perhitungan:

   $$\mathrm{RMSE}=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(\rho_{i,\mathrm{ukur}}-\rho_{i,\mathrm{model}})^2}$$

4. Hitung koefisien determinasi $R^2$ dengan metode yang dinyatakan secara eksplisit dan sesuai tujuan evaluasi.
5. Buat cross-plot antara resistivitas pengukuran dan resistivitas model.
6. Buat visualisasi residual untuk memperlihatkan pola kesalahan.
7. Evaluasi apakah pola anomali konduktif yang teridentifikasi pada masing-masing metode saling mendukung.
8. Jika tersedia data Total Dissolved Solids (TDS), Daya Hantar Listrik (DHL), atau data hidrokimia, gunakan data tersebut sebagai pembanding independen.
9. Akhiri dengan kesimpulan terpadu berdasarkan kekuatan bukti dan keterbatasan data.

**Context**

Investigasi geolistrik di TPA Tamangapa bertujuan mengevaluasi indikasi sebaran air lindi berdasarkan variasi resistivitas bawah permukaan.

Konsistensi hasil antar-metode dapat memperkuat interpretasi, tetapi tidak otomatis membuktikan pencemaran.

Validasi statistik perlu didasarkan pada pasangan data yang benar-benar sebanding, sedangkan konfirmasi kondisi air tanah memerlukan data pendukung, seperti pengukuran hidrokimia atau sampel air.

**Input**

Gunakan hasil perhitungan resistivitas dari metode Wenner–Schlumberger, Wenner, VES, dan Dipole–Dipole yang telah dibuat pada tahap sebelumnya.

Gunakan data pengukuran dan model yang berpasangan apabila tersedia.

Jika menggunakan data simulasi, nyatakan bahwa evaluasi tersebut merupakan validasi internal terhadap data simulasi, bukan validasi lapangan independen.

Gunakan data TDS dan DHL hanya apabila benar-benar tersedia. Jangan membuat data pengukuran atau mengklaim validasi hidrokimia jika tidak ada.

**Constraints**

1. Hitung residual menggunakan $e_i=\rho_{i,\mathrm{ukur}}-\rho_{i,\mathrm{model}}$.
2. Hitung RMSE berdasarkan data aktual yang digunakan.
3. Hitung $R^2$ dengan metode yang dinyatakan secara eksplisit dan sesuai tujuan evaluasi.
4. Jangan menyamakan koefisien korelasi dengan koefisien determinasi.
5. Jangan menetapkan nilai $R^2$, RMSE, atau persentase error secara manual tanpa menghitungnya dari data.
6. Jangan menyebut analisis sebagai validasi independen jika data model dan data pembanding berasal dari sumber yang sama atau dibuat secara simulasi.
7. Jangan mengarang data TDS, DHL, koordinat, atau parameter hidrokimia.
8. Gunakan Python dan Matplotlib untuk menghasilkan grafik dan evaluasi numerik.

**Quality Criteria**

- Seluruh metrik statistik harus dapat direproduksi dari data input.
- Grafik harus memiliki sumbu, satuan, legenda, dan garis referensi yang tepat.
- Analisis residual harus menjelaskan besar serta pola perbedaan antara data pengukuran dan model.
- Kesimpulan akhir harus membedakan konsistensi antar-metode, tingkat ketidakpastian, dan bukti independen yang benar-benar tersedia.

**Output Akhir**

1. Tabel perbandingan resistivitas pengukuran dan model, termasuk residual.
2. Cross-plot resistivitas pengukuran versus model dengan garis referensi 1:1.
3. Nilai $R^2$ dan RMSE hasil perhitungan, apabila layak dihitung.
4. Grafik atau peta residual berdasarkan informasi spasial yang benar-benar tersedia.
5. Kesimpulan integrasi empat metode dan evaluasi keterbatasan interpretasi pencemaran air lindi.

## **D. Kesimpulan**

Penggunaan metode geolistrik resistivitas melalui lima prompt memberikan alur analisis yang sistematis untuk mempelajari variasi resistivitas bawah permukaan dalam konteks identifikasi indikasi sebaran lindi (*leachate*) di TPA Tamangapa, Antang, Kota Makassar. Kelima prompt dirancang untuk saling melengkapi, mulai dari pemodelan penampang resistivitas dua dimensi, analisis variasi lateral, interpretasi perlapisan vertikal, visualisasi distribusi resistivitas tiga dimensi, hingga integrasi dan evaluasi hasil antar-metode.

Metode Wenner–Schlumberger digunakan untuk menggambarkan distribusi resistivitas bawah permukaan secara dua dimensi. Metode Wenner berfokus pada variasi resistivitas lateral, sedangkan metode Schlumberger melalui *Vertical Electrical Sounding* (VES) digunakan untuk mengkaji perubahan resistivitas terhadap kedalaman. Sementara itu, metode Dipole–Dipole membantu menggambarkan variasi resistivitas lateral dan mendukung visualisasi distribusi tiga dimensi apabila data yang tersedia memadai. Pada prompt kelima, hasil dari berbagai metode diintegrasikan untuk menilai konsistensi interpretasi dan mengevaluasi kualitas model menggunakan parameter yang sesuai dengan data.

Secara keseluruhan, kelima prompt membentuk rangkaian kerja yang terstruktur dan mendukung pengolahan, perhitungan, visualisasi, serta interpretasi data resistivitas menggunakan Python. Pendekatan ini dapat membantu mengidentifikasi zona dengan nilai resistivitas relatif rendah yang berpotensi berkaitan dengan material jenuh air atau keberadaan lindi. Namun, nilai resistivitas rendah tidak dapat dijadikan satu-satunya dasar untuk menyatakan adanya pencemaran karena karakteristik tersebut juga dapat dipengaruhi oleh jenis batuan, kandungan lempung, porositas, dan kondisi kejenuhan air.

Dengan demikian, penggunaan kelima prompt dapat menjadi kerangka analisis awal untuk memahami karakteristik bawah permukaan dan mengevaluasi potensi sebaran lindi di sekitar TPA Tamangapa. Keandalan interpretasi tetap bergantung pada kualitas dan jenis data masukan, kesesuaian konfigurasi elektroda, serta ketepatan pemodelan. Jika analisis menggunakan data simulasi, hasil yang diperoleh hanya menunjukkan penerapan metode dan alur analisis, bukan bukti langsung adanya pencemaran di lokasi penelitian. Oleh karena itu, verifikasi melalui data geolistrik lapangan dan data pendukung hidrogeologi atau kualitas air diperlukan untuk memperkuat interpretasi.
